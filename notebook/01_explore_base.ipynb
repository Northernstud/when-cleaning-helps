### This file is for exploring the data set

```python
# import pandas
import pandas as pd 

# get the file path
drug_consumption = "../data/raw/Drug_consumption.csv"

# read the dataset
df = pd.read_csv(drug_consumption)

# create the dataset from df that include only categorical data
categorical_variables = df.select_dtypes(include=['object', 'category'])

print(f"The shape of the dataset: {df.shape}.") # print the numbers of rows and columns
print(f"The columns of the dataset: {df.columns}.") # print the name of the columns

# creating the dataframe with rows as drug_consumption's columns and columns as unique and missing values
summary_df = pd.DataFrame({
    'Unique Values': categorical_variables.nunique(),
    'Missing Values': categorical_variables.isnull().sum()
})

#use the display.max_rows, none to make display show the whole liest
with pd.option_context("display.max_rows", None):
    display(summary_df)

# display first 5 rows
with pd.option_context("display.max_columns", None):
    display(df)

df["Nicotine"].value_counts()
```
